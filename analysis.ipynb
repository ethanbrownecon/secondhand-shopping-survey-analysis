# Second-hand shopping survey analysis

Explore purchasing behavior, shopping timing, motivations, and willingness to pay. The respondent-level source file is intentionally not included.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

survey_path = Path("data/secondhand_shopping_survey.csv")
surveyData = pd.read_csv(survey_path)
surveyData = surveyData.drop(columns=["Timestamp", "Name", " "])
surveyData.head()
```

# 1ST WHO - Age

```python

plt.rcParams['figure.figsize'] = [10, 6]
ageGroups = surveyData.groupby('Age')

plt.pie(ageGroups.size(), labels=ageGroups.size().index, autopct='%1.1f%%', colors= ['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6'])
plt.colormaps
plt.title('Age Distribution of Survey Participants')
plt.show()


```

# 2ND WHO - Gender

```python


genderGroups = surveyData.groupby('Gender')

plt.pie(genderGroups.size(), labels=genderGroups.size().index, autopct='%1.1f%%')
plt.title('Gender Distribution of Survey Participants')
plt.show()
```

# 3RD WHO - Year 

```python


yearGroups = surveyData.groupby('Year')

plt.pie(yearGroups.size(), labels=yearGroups.size().index, autopct='%1.1f%%')
plt.title('Year Distribution of Survey Participants')
plt.show()
```

# 3RD WHEN
# What days are you most likely to shop for secondhand  clothing?

```python

Monday = surveyData['What days are you most likely to shop for secondhand  clothing?'].str.contains("Monday", na=False)
Tuesday = surveyData['What days are you most likely to shop for secondhand  clothing?'].str.contains("Tuesday", na=False)
Wednesday = surveyData['What days are you most likely to shop for secondhand  clothing?'].str.contains("Wednesday", na=False)
Thursday = surveyData['What days are you most likely to shop for secondhand  clothing?'].str.contains("Thursday", na=False)
Friday = surveyData['What days are you most likely to shop for secondhand  clothing?'].str.contains("Friday", na=False)
Saturday = surveyData['What days are you most likely to shop for secondhand  clothing?'].str.contains("Saturday", na=False)
Sunday = surveyData['What days are you most likely to shop for secondhand  clothing?'].str.contains("Sunday", na=False)

total_days_responses = surveyData['What days are you most likely to shop for secondhand  clothing?'].count()

print("Monday:", Monday.sum() / total_days_responses * 100)
print("Tuesday:", Tuesday.sum() / total_days_responses * 100)
print("Wednesday:", Wednesday.sum() / total_days_responses * 100)
print("Thursday:", Thursday.sum() / total_days_responses * 100)
print("Friday:", Friday.sum() / total_days_responses * 100)
print("Saturday:", Saturday.sum() / total_days_responses * 100)
print("Sunday:", Sunday.sum() / total_days_responses * 100)


# create a bar chart for the days of the week
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
percentages = [Monday.sum() / total_days_responses * 100,
               Tuesday.sum() / total_days_responses * 100,
               Wednesday.sum() / total_days_responses * 100,
               Thursday.sum() / total_days_responses * 100,
               Friday.sum() / total_days_responses * 100,
               Saturday.sum() / total_days_responses * 100,
               Sunday.sum() / total_days_responses * 100]       
plt.bar(days, percentages, color=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6','#c4e17f'])
plt.xlabel('Days of the Week')
plt.ylabel('Percentage of Respondents (%)')
plt.title('Preferred Days for Secondhand Clothing Shopping')
plt.show()
```

# MAIN WHERE QUESTION
# Where do you most often shop for secondhand clothing?

```python


ThriftStores = surveyData['Where do you most often shop for secondhand clothing?'].str.contains("Thrift stores")
FleaMarkets = surveyData['Where do you most often shop for secondhand clothing?'].str.contains("Flea Markets")
OnlineStores = surveyData['Where do you most often shop for secondhand clothing?'].str.contains("Online stores")
MarketsEvents = surveyData['Where do you most often shop for secondhand clothing?'].str.contains("Markets/Events")
VintageShops = surveyData['Where do you most often shop for secondhand clothing?'].str.contains("Vintage shops")

total_responses = surveyData['Where do you most often shop for secondhand clothing?'].count()

print("ThriftStores:", ThriftStores.sum() / total_responses * 100)
print("FleaMarkets:", FleaMarkets.sum() / total_responses * 100)
print("OnlineStores:", OnlineStores.sum() / total_responses * 100)
print("MarketsEvents:", MarketsEvents.sum() / total_responses * 100)
print("VintageShops:", VintageShops.sum() / total_responses * 100)
#create bar chart 
places = ['ThriftStores', 'FleaMarkets', 'OnlineStores', 'MarketsEvents', 'VintageShops']
percentages_places = [ThriftStores.sum() / total_responses * 100,
                        FleaMarkets.sum() / total_responses * 100,
                        OnlineStores.sum() / total_responses * 100,
                        MarketsEvents.sum() / total_responses * 100,
                        VintageShops.sum() / total_responses * 100]
plt.bar(places, percentages_places, color=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0'])
plt.xlabel('Places to Shop for Secondhand Clothing')
plt.ylabel('Percentage of Respondents (%)')
plt.title('Preferred Places for Secondhand Clothing Shopping')
plt.show()
```

# WHY THE WHAT
# Why do you typically choose secondhand clothing?



```python
plt.rcParams['figure.figsize'] = [15, 6]

UniqueStyle = surveyData['Why do you typically choose secondhand clothing?'].str.contains('Unique style')
Sustainability = surveyData['Why do you typically choose secondhand clothing?'].str.contains('Sustainability / reusing clothing')
BetterQuality = surveyData['Why do you typically choose secondhand clothing?'].str.contains('Better quality or durability')
LowerCost = surveyData['Why do you typically choose secondhand clothing?'].str.contains('Lower cost')
EnvironmentalConcerns = surveyData['Why do you typically choose secondhand clothing?'].str.contains('Environmental concerns')
Other = surveyData['Why do you typically choose secondhand clothing?'].str.contains('Other:')

total_responses = surveyData['Why do you typically choose secondhand clothing?'].count()

print("Unique Style:", UniqueStyle.sum() / total_responses * 100)
print("Sustainability / reusing clothing:", Sustainability.sum() / total_responses * 100)
print("Better quality or durability:", BetterQuality.sum() / total_responses * 100)
print("Lower cost:", LowerCost.sum() / total_responses * 100)
print("Environmental concerns:", EnvironmentalConcerns.sum() / total_responses * 100)
print("Other:", Other.sum() / total_responses * 100)
#bar chart
reasons_secondhand = ['Unique Style', 'Sustainability', 'Better quality/durability', 'Lower cost', 'Environmental concerns', 'Other']
percentages_secondhand = [UniqueStyle.sum() / total_responses * 100,
                        Sustainability.sum() / total_responses * 100,
                        BetterQuality.sum() / total_responses * 100,
                        LowerCost.sum() / total_responses * 100,
                        EnvironmentalConcerns.sum() / total_responses * 100,
                        Other.sum() / total_responses * 100]
plt.bar(reasons_secondhand, percentages_secondhand, color=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6'])
plt.xlabel('Reasons for Choosing Secondhand Clothing')
plt.ylabel('Percentage of Respondents (%)')
plt.title('Reasons for Choosing Secondhand Clothing')
plt.show()
```


# Average willingness to pay for secondhand clothing

```python



#What is the most you would feel comfortable paying for a T-shirt at my booth?
#What is the most you would feel comfortable paying for a outerwear(light jackets/flannels/long sleeves) at my booth?
#What is the most you would feel comfortable paying for a pair of pants at my booth?
#What is the most you would feel comfortable paying for a hat at my booth?

#drop if contains non-numeric values (~ or $ or any strings)
#drop all other columns except for the price columns
willingnesstopay = surveyData[['What is the most you would feel comfortable paying for a T-shirt at my booth?', 
                              'What is the most you would feel comfortable paying for a outerwear(light jackets/flannels/long sleeves) at my booth?',
                              'What is the most you would feel comfortable paying for a pair of pants at my booth?',
                              'What is the most you would feel comfortable paying for a hat at my booth?']]


willingnesstopay = willingnesstopay.replace('[$,~,depending on quality,depending on what kind (sweats jeans etc)]', '', regex=True)
willingnesstopay = willingnesstopay.astype(float)


average_tshirt_price = willingnesstopay['What is the most you would feel comfortable paying for a T-shirt at my booth?'].mean()
average_outerwear_price = willingnesstopay['What is the most you would feel comfortable paying for a outerwear(light jackets/flannels/long sleeves) at my booth?'].mean()
average_pants_price = willingnesstopay['What is the most you would feel comfortable paying for a pair of pants at my booth?'].mean()
average_hat_price = willingnesstopay['What is the most you would feel comfortable paying for a hat at my booth?'].mean()
                                     
print("Average T-shirt Price: $", average_tshirt_price)
print("Average Outerwear Price: $", average_outerwear_price)
print("Average Pants Price: $", average_pants_price)
print("Average Hat Price: $", average_hat_price)


#tshirt = surveyData['What is the most you would feel comfortable paying for a T-shirt at my booth?'].astype(float)

# bar chart
items = ['T-shirt', 'Outerwear', 'Pants', 'Hat']
average_prices = [average_tshirt_price,
                    average_outerwear_price,    
                    average_pants_price,
                    average_hat_price]
plt.bar(items, average_prices, color=['#ff9999','#66b3ff','#99ff99','#ffcc99']) 
plt.xlabel('Clothing Items')
plt.ylabel('Average Willingess To Pay Price ($)')
plt.title('Average Willingness To Pay for Secondhand Clothing Items at My Booth')
plt.show()
```
